# Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable while values can be of any type.

```python
## Creating Dictionaries

empty_dict = {}
print(empty_dict)
print(type(empty_dict))
```

```python
empty_dict=dict()
print(empty_dict)
```

```python
student={"name":"Ben","age":45,"grade":"A"}
print(student)
print(type(student))
```

```python
student = {'name': 'Harry', 'age': 24, 'name':'Potter' }
print(student)
```

```python
## Accessing Elements in Dictionaries
student = {'name': 'Harry', 'age': 24, "grade":"A" }
print(student['name'])
print(student['age'])
print(student['grade'])
```

```python
## Accessing using get() method
print(student.get('name'))
print(student.get('age'))
```

```python
print(student.get('last_name')) 
print(student.get('last_name','Not Found')) 
```

```python
## Modifyinf dictionary elements
## Dictionaries are mutable
student['age']=25 ## Updated age
print(student)
student['address']='Hogwarts' ## added new key-value pair
print(student)
```

```python
del student['grade'] ## Deleting a key-value pair
print(student)
```

```python
## Dictionary methods
keys=student.keys()
print(keys)
values=student.values()
print(values)
items=student.items()
print(items)
```

```python
## shallow copy
student_copy=student
print(student_copy)
print(student)
student["name"]="Hermione"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## deep copy
print(student_copy1)
student["name"]="Ron"
print(student)
```

```python
## Iterating over dictionaries
## We can use loops to iterate over dictionaries, keys, values or items
for keys in student.keys():
    print(keys)

```

```python
## Iterate over values

for values in student.values():
    print(values)
```

```python
## Iterate over key-value pairs
for key,value in student.items():
    print(f"Key is {key} and value is {value}")
```

```python
## Nested dictionaries
student={
    "student1":{"name":"Harry","age":32},
    "student2":{"name":"Ron","age":28},

}
print(student)
```

```python
## Access nested dictionary elements
print(student['student1']['name'])
print(student['student2']['age'])
```

```python
### Iterating over nested dictionaries
for student_id,student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Condition dictionary comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Merging 2 dictionaries
dict1={'a':1,'b':2}
dict2={'c':3,'d':4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
