# The map() function in Python

The map() function applies a given function to all items in an input list or any other iterable and returns a map object(an iterator). This is particularly useful for transforming a data in a list comprehensively.

```python
def square(x):
    return x**2
square(10)
```

```python
numbers=[1,2,3,4,5,6,7,8]
list(map(square,numbers))
```

```python
## Lambda function with map

numbers=[1,2,3,4,5,6,7,8]
list(map(lambda x:x**2,numbers))
```

```python
## Map multiple iterables
numbers1=[1,2,3]
numbers2=[4,5,6]
added_numbers=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
## map to convert a list of strings to integers

str_numbers=['1','2','3','4','5']
int_numbers=list(map(int,str_numbers))
print(int_numbers)
```

```python
words=['apple','banana','cherry']
upper_word=list(map(str.upper,words))
print(upper_word)
```
