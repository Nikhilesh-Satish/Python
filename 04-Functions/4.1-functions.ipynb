```python
def even_or_odd(num):
    """This function finds even or odd""" ## Docstring
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
```

```python
## Call this function
even_or_odd(5)
```

```python
## function with multiple parameters
def add(a,b):
    return a+b
result=add(2,4)
print(result)
```

```python
## Default parameters
def greet(name="Guest"):
    print(f"Hello {name}")
greet("Malfoy")
greet()
```

```python
## Variable length arguments
## Positional and keywords arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5)
```

```python
### Keywords arguments- All parameters are key-value pairs
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Draco",age=452,city="Azkaban")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional argument : {val}")
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,name="Harry",age=23) ## Positional arguments cannot come after keywords arguments
```
