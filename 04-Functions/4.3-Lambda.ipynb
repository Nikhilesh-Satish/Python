# Lambda functions in Python

Lambda functions are small anonymous functions defined using lambda keyword. Can have multiple arguments but only 1 expression. Can be used for short operations or as arguments to higher order functions

Syntax

    lambda arguments : expression

```python
def addition(a,b):
    return a+b
```

```python
addition(2,3)
```

```python
addition=lambda a,b:a+b
type(addition)
```

```python
print(addition(2,3))
```

```python
def even(num):
    if num%2==0:
        return True
    else:
        return False
even(4)
```

```python
even1=lambda num:num%2==0
even1(4)
```

```python
## map()

numbers=[1,2,3,4,5,6]
def square(num):
    return num**2
print(square(2))
```

```python
## map()-applies a function to all itema in a list
list(map(lambda x:x**2,numbers)) ## lazy function, does not execute immediately
```
