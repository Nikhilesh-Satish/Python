## SQL and SQLite

SQLite is a self-contained, serverless and zero-configuration db engine that is widely used for embedded database systems. 

```python
import sqlite3
```

```python
## Connect to an SQLite database

connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
## Create a table
cursor.execute('''
               Create Table if not exists employees(
               id Integer Primary Key,
               name Text Not Null,
               age Integer,
               department text)''')
connection.commit()
```

```python
cursor.execute('''Select * from employees''')
```

```python
## Insert data in sqlite table

cursor.execute('''Insert into employees (name, age, department) values('Arteta',42,'Manager')''')
cursor.execute('''Insert into employees (name, age, department) values('Saka',24,'Winger')''')
cursor.execute('''Insert into employees (name, age, department) values('Gyokeres',27,'Striker')''')

connection.commit()
```

```python
## Querying the data from the table

cursor.execute('Select * from employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
## Update the data in the table

cursor.execute('''Update employees set age=34 where name='Arteta' ''')
connection.commit()
```

```python
cursor.execute('Select * from employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
## Delete the data from the table
cursor.execute('''Delete from employees where name='Gyokeres' ''')
connection.commit()
```
