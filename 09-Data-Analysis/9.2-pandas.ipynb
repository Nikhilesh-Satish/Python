## Pandas - Daataframe And Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A series is a 1-D array-like object, while a DataaFrame is a 2-Dimensional, size mutable and potentially heterogeneous tablular data structures with labeled axes (rows and columns)

```python
!pip install pandas
import pandas as pd
```

```python
## Series
## A pandas series is a one dimensional arrat-like object that can hold various data types. It is similar to a column in a table

data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
```

```python
## Create a series from dictionary
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
## Dataframe - Multiple rows and columns

## create a dataframe from dictionary of list
data={
    'Name':['Arteta','Emery','Wenger'],
    'Age':[42,50,71],
    'City':['London','Manchester','Paris']
}
df=pd.DataFrame(data)
print(df)   
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## Create a Data frame from a list of dictionaries

data=[
    {'Name':'Arteta','Age':42,'City':'London'},
    {'Name':'Emery','Age':50,'City':'Manchester'},
    {'Name':'Wenger','Age':71,'City':'Paris'},
    {'Name':'Graham','Age':92,'City':'London'}

]

df=pd.DataFrame(data)
print(df)
print(type(df))
```
