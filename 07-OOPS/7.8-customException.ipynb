## Custom Exception (Raise and throw an exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass

```

```python
year=int(input("Enter the year of birth"))
age=2025-year

try:
    if age<=30 and age>=20:
        print("You are eligible")
    else:
        raise dobException
except dobException:
    print("You are not eligible")
```
