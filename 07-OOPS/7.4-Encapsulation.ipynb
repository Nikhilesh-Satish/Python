## Encapsulation

- Encapsulation is the concept of wrapping data and methods together as a single unnit. It restricts direct access to some of the object's components which is a means of preventing accidental interference and misuse of data



```python
## Encapsulation with getter and setter methods
## Public, protected and private variables or access modifiers
class Person:
    def __init__(self,name,age,gender,salary):
        self.__name = name ##private variable        
        self.__age=age  ##private variable
        self.gender=gender
        self._salary=salary  ##protected variable
def get_name(person):
    return person.__name

person=Person("Alice",30,"Female")
## get_name(person) - Throws error because __name is private variable
```

```python
## Encapsulation with getter and setter methods
class Person:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
    def get_name(self):
        return self.__name
    def set_name(self,name):
        self.__name=name
```
