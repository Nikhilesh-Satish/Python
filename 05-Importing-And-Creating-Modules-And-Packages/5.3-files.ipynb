## File operation-Read and write files

Python provides built in functions and methods to read from and write to files, both text and binary.

```python
### Read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
## Read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() removes newline character
```

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:
    file.write("This is a new line.\n")
    file.write("This will overwrite the existing content.\n")

```

```python
## Write a file without overwriting
with open('example.txt','a') as file:
    file.write("This line is appended.\n")
    file.write("This will not overwrite the existing content.\n")
```

```python
### Writing a list of lines to a file
lines = ["First line\n", "Second line\n", "Third line\n"]
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
### Binary files
data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
### Reading a binary file
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

```python
## File paths
## Listing files and directories
import os
items=os.listdir('.')
print(items)
```

```python
### Joining paths
dir_name="folder"
file_name="file.txt"
full_path=os.path.join(dir_name,file_name)
print(full_path)
```
